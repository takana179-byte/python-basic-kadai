```python
# 元のデータ
product_name = "ワイヤレスマウス"
price = "3480"
count = "3"
tax_rate = "0.1"

# 計算
price_num = int(price)
count_num = int(count)
tax_num = float(tax_rate)

zeinuki = price_num * count_num
shouhizei = int(zeinuki * tax_num)
zeikomi = zeinuki + shouhizei

# print
print("商品名： " + product_name)
print("単価： " + str(price_num) + " 円")
print("購入数： " + str(count_num) + " 個")
print("税抜価格： " + str(zeinuki) + " 円")
print("消費税： " + str(shouhizei) + " 円")
print("税込価格： " + str(zeikomi) + " 円")

# 確認
print(type(price_num))
print(type(count_num))
print(type(tax_num))
```
