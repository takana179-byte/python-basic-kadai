```python
import math

#データ
radius = 7

# 計算
diameter = radius * 2
circumference = 2 * math.pi * radius
area = math.pi * (radius ** 2)
area_sqrt = math.sqrt(area)
area_ceil = math.ceil(area)
area_floor = math.floor(area)

# 出力
print("半径： " + str(radius))
print("直径： " + str(diameter))
print("円周： " + str(circumference))
print("面積： " + str(area))
print("面積の平方根： " + str(area_sqrt))
print("面積の切り上げ： " + str(area_ceil))
print("面積の切り捨て： " + str(area_floor))
```
