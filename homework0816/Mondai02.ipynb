```python
#データ
scores = [72, 91, 68, 85, 79]

# 計算
max_score = max(scores)
min_score = min(scores)
total_score = sum(scores)
count = len(scores)
avg_score = total_score / count
sorted_scores = sorted(scores)

# 出力
print("最高点： " + str(max_score) + " 点")
print("最低点： " + str(min_score) + " 点")
print("合計点： " + str(total_score) + " 点")
print("平均点： " + str(avg_score) + " 点")
print("テスト回数： " + str(count) + " 回")
print("点数順： " + str(sorted_scores))

```
