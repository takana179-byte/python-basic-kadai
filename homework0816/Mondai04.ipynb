```python
#Data
employee = {
    "name": "田中太郎",
    "age": 28,
    "department": "開発部",
    "salary": 320000
}

#1
print("### 1の結果を出力")
print("氏名： " + employee["name"])
print("年齢： " + str(employee["age"]))
print("所属： " + employee["department"])
print("月給： " + str(employee["salary"]))

#2
employee["age"] = 29

#3
employee["department"] = "AI開発部"

#4
employee["bonus"] = 500000

#2-4を行った結果の連想配列を出力
print("###2-4を行った結果の連想配列を出力")
print(employee)

# 5. 年間給与の計算（月給12か月分 ＋ ボーナス）
annual_salary = (employee["salary"] * 12) + employee["bonus"]

# 年間給与の出力
print("###5の結果を出力")
print("年間給与： " + str(annual_salary) + " 円")
```
