```python
# 1. Humanクラスの定義
class Human:

    def __init__(self, name, age):
        self.name = name  # 名前（属性）
        self.age = age  # 年齢（属性）

    # 大人かどうか判定して出力するメソッド
    def check_adult(self):
        if self.age >= 20:
            print(self.name + "さんは大人です。")
        else:
            print(self.name + "さんは大人ではありません。")


# 2. インスタンスの生成
human1 = Human("侍太郎", 25)
human2 = Human("侍花子", 18)
human3 = Human("侍一郎", 20)

# 3. インスタンスをリストに追加
human_list = [human1, human2, human3]

# 4. リストの要素数分だけ繰り返し処理（for文）を行い、check_adultメソッドを呼び出す
for human in human_list:
    human.check_adult()
```
