```python
# Humanクラスの定義
class Human:
    # 初期化メソッド（インスタンス化するときに属性を設定する）
    def __init__(self, name, age):
        self.name = name  # 名前（属性）
        self.age = age    # 年齢（属性）

    # 名前と年齢を出力するメソッド
    def printinfo(self):
        print("名前：" + self.name)
        print("年齢：" + str(self.age) + "歳")


# Humanクラスのインスタンスを作成し、変数 human に代入
human = Human("侍太郎", 20)

# 定義したメソッド（printinfo）を呼び出して出力
human.printinfo()
```
